```python
import pandas as pd
from sklearn.ensemble import RandomForestClassifier, RandomForestRegressor
from mlxtend.feature_selection import SequentialFeatureSelector as SFS
```

# Dataset preparation

```python
clinical_df = pd.read_csv(r"D:\Clinical_data\Table_clinical_data.csv", encoding="cp1252")
df_vmi40 = pd.read_csv(r"D:\CSV\merged\merged_monoe_40kev_radiomics_spine_lesions_features.csv")

pd.set_option('future.no_silent_downcasting', True)
clinical_df['Stage'] = clinical_df['ISS classification'].replace({'Stage 1': 1,'Stage 2': 2,'Stage 3': 3})
clinical_df = clinical_df.dropna(subset=['Stage'])
clinical_df['Stage'] = clinical_df['Stage'].astype(int)

X = df_vmi40.select_dtypes(include=['number'])

y = clinical_df['Stage']
# y = clinical_df['Beta2 microglobulin (mg/l)']

valid_idx = X.notna().all(axis=1) & y.notna()
X = X[valid_idx]
y = y[valid_idx]
```

# Floating forward selection

Classification

```python
model = RandomForestClassifier(
    n_estimators=200,
    random_state=42,
    n_jobs=1
)

sfs_floating_forward = SFS(
    model,
    k_features=20,
    forward=True,
    floating=True,
    scoring='accuracy',
    cv=5,
    n_jobs=-1
)

sfs_forward = sfs_floating_forward.fit(X, y)
```

```python
selected_features_forward = list(sfs_forward.k_feature_names_)
print(selected_features_forward)
```

Regression

```python
model = RandomForestRegressor(
    n_estimators=200,
    random_state=42,
    n_jobs=1
)

sfs_floating_forward = SFS(
    model,
    k_features=20,
    forward=True,
    floating=True,
    scoring='neg_mean_absolute_error',
    cv=5,
    n_jobs=-1
)

sfs_forward = sfs_floating_forward.fit(X, y)
```

```python
selected_features_forward = list(sfs_forward.k_feature_names_)
print(selected_features_forward)
```
