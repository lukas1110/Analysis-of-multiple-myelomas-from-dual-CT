```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv(r"D:\CSV\merged\merged_CaSupp_25_radiomics_spine_lesions_features.csv")
```

```python
# BOXPLOT
columns = ["n_lesions", "original_shape_MeshVolume", "original_shape_SurfaceArea", "original_shape_SurfaceVolumeRatio", "original_shape_VoxelVolume"]
n_cols = len(columns)
cols_per_row = 5
rows = (n_cols + cols_per_row - 1) // cols_per_row

plt.figure(figsize=(cols_per_row*3, rows*3))
for i, col in enumerate(columns):
    plt.subplot(rows, cols_per_row, i+1)
    plt.boxplot(df[col].dropna())
    plt.title(col, fontsize=10)
    plt.xticks([])
plt.tight_layout()
plt.show()
```

```python
# HISTOGRAMS
plt.figure(figsize=(cols_per_row*3, rows*3))
for i, col in enumerate(columns):
    plt.subplot(rows, cols_per_row, i+1)
    plt.hist(df[col].dropna(), bins=20, color='skyblue', edgecolor='black')
    plt.title(col, fontsize=10)
    plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
